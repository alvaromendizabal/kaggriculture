### 2. Registered comparison

The new route descriptor subtracts the best alternatives other friendly workers would lose if this route consumes their resources. It affects **menu admission only**. Each incumbent assignment route and direct deposit are protected, with eight non-PASS choices still allowed per worker. The same exact allocator and objective are used. A static tie keeps the old assignment.

Preserving the incumbent prevents a worse static objective, **not** a worse realized game outcome. The pilot therefore measures continuously advanced trajectories under a responsive opponent.

```python
protocol = json.loads((BASE / 'PROTOCOL.json').read_text())
display(pd.DataFrame([{'screen_states':69, 'screen_cap_seconds':90,
    'pilot_cap_seconds':120, 'callback_cap_ms':500,
    'max_primary_pairs':1, 'max_negative_controls':1,
    'max_research_transitions':92, 'new_full_seasons':0,
    'non_PASS_menu_slots_per_worker':8}]))
print('Selection: first activated development source by seed/seat order, never by endpoint outcome.')
```

```python
def run_stage(stage):
    emergency_seconds = {'screen':110, 'pilot':140}[stage]
    cmd = [sys.executable, str(BASE / 'run_routes.py'), stage]
    proc = subprocess.Popen(cmd, cwd=BASE, stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT, text=True, bufsize=1, start_new_session=True)
    started = time.monotonic()
    try:
        while proc.poll() is None:
            if time.monotonic() - started > emergency_seconds:
                raise TimeoutError('Notebook emergency deadline; bundle diagnostics.')
            ready, _, _ = select.select([proc.stdout], [], [], 1.0)
            if ready:
                line = proc.stdout.readline()
                if line: print(line, end='', flush=True)
        tail = proc.stdout.read()
        if tail: print(tail, end='')
        if proc.returncode:
            raise RuntimeError('Stage failed. Stop and use the bundle command; do not retry unchanged.')
    except BaseException:
        if proc.poll() is None:
            os.killpg(proc.pid, signal.SIGTERM)
            try: proc.wait(timeout=5)
            except subprocess.TimeoutExpired:
                os.killpg(proc.pid, signal.SIGKILL)
                proc.wait()
        raise
```
