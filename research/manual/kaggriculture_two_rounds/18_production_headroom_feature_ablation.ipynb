```python
from pathlib import Path
import json, os, signal, subprocess, sys, threading, queue, time
import pandas as pd
import plotly.io as pio
from IPython.display import display
BASE = Path.cwd()
if not (BASE / "run_research.py").is_file():
    BASE = Path.home() / "kaggriculture_two_rounds"
if not (BASE / "run_research.py").is_file():
    raise FileNotFoundError("Open this notebook inside the extracted two-round package.")
sys.path.insert(0, str(BASE))
from visualize import figures, dashboard
pio.renderers.default = "plotly_mimetype"
ROUND = "18"
print("LIVE AWS EXECUTION • round", ROUND)
print("Package:", BASE)
print("Interpreter:", sys.executable)

def run_stage(command, block=None):
    args = [sys.executable, str(BASE / "run_research.py"), command, "--round", ROUND]
    if block is not None:
        args += ["--block", str(block)]
    cap = 150 if command == "screen" else 210 if command == "pair" else 45
    proc = subprocess.Popen(args, cwd=BASE, stdout=subprocess.PIPE,
                            stderr=subprocess.STDOUT, text=True, start_new_session=True)
    messages = queue.Queue()
    def reader():
        for line in proc.stdout:
            messages.put(line)
    threading.Thread(target=reader, daemon=True).start()
    started = time.monotonic()
    try:
        while proc.poll() is None or not messages.empty():
            try:
                print(messages.get(timeout=0.2), end="", flush=True)
            except queue.Empty:
                pass
            if time.monotonic() - started > cap:
                raise TimeoutError("Notebook emergency deadline; preserve diagnostics.")
        if proc.returncode:
            raise RuntimeError("Stage failed. Save the notebook and run the bundle command; do not retry unchanged.")
    finally:
        if proc.poll() is None:
            os.killpg(proc.pid, signal.SIGTERM)
            try:
                proc.wait(timeout=5)
            except subprocess.TimeoutExpired:
                os.killpg(proc.pid, signal.SIGKILL)
                proc.wait()
```
