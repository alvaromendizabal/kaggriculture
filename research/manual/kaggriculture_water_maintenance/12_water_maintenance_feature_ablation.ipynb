## 3 · Run the bounded saved-observation screen

This stage verifies package/dependency fingerprints, the reviewed checkout, the pinned engine, and all ten restored objects. It does not rerun earlier notebooks.

Before any research observation is processed, **736 artificial crop fixtures** compare the standalone scenario calculation against the installed official WATER, decay, and nightly-refresh primitives, in both WATER/no-WATER paths. These are correctness fixtures, not games or performance results.

The screen samples 468 states from the three preserved coordinated episodes, including every hour in the intervention window. Only changed actions inside the registered window can activate the pilot. Source selection is the first activated seed/seat, never a choice based on rewards.

The stage prints UTC heartbeats, saves each completed source, and rejects an unchanged failed or interrupted restart.

````python
receipt = Path.home() / "kaggriculture_manual_resume/state/runtime.json"
if not receipt.exists():
    raise FileNotFoundError("Existing runtime receipt is missing. Do not reinstall; bundle the error.")
PYTHON_BIN = json.loads(receipt.read_text())["executable"]
if Path(sys.executable).absolute() != Path(PYTHON_BIN).absolute():
    raise RuntimeError("Select Kernel → Kaggriculture Manual (verified source), then restart this notebook.")

# Streams child logs without blocking the notebook's emergency deadline.
# The runner owns its inner worker's hard cap and termination.
def run_stage(stage):
    caps = {"screen": 120, "pilot": 180}
    if stage not in caps:
        raise ValueError("Unregistered stage")
    OUT.mkdir(exist_ok=True)
    logpath = OUT / (stage + "_notebook_console.txt")
    process = None
    offset = 0
    started = time.monotonic()
    def show_new():
        nonlocal offset
        with logpath.open("r", errors="replace") as stream:
            stream.seek(offset)
            text = stream.read()
            offset = stream.tell()
        if text:
            print(text, end="", flush=True)
    try:
        with logpath.open("w") as console:
            process = subprocess.Popen(
                [PYTHON_BIN, "-u", str(ROOT / "run_maintenance.py"), stage],
                cwd=ROOT, stdout=console, stderr=subprocess.STDOUT,
                start_new_session=True,
            )
            while process.poll() is None:
                show_new()
                if time.monotonic() - started > caps[stage] + 30:
                    raise TimeoutError("Notebook emergency deadline; bundle diagnostics.")
                time.sleep(0.25)
            show_new()
            if process.returncode:
                raise RuntimeError(f"{stage} failed. Stop; save and bundle. See {logpath}.")
    except BaseException:
        if process is not None and process.poll() is None:
            os.killpg(process.pid, signal.SIGTERM)
            try:
                process.wait(timeout=5)
            except subprocess.TimeoutExpired:
                os.killpg(process.pid, signal.SIGKILL)
                process.wait()
        raise
    return json.loads((OUT / stage / "report.json").read_text())
````

````python
screen = run_stage("screen")
display(pd.Series({k:v for k,v in screen.items() if k not in ("output_sha256",)}, name="live screen result").to_frame())
mechanics = json.loads((OUT / "mechanics.json").read_text())
print("Installed-engine component checks:", mechanics["status"], mechanics["fixtures"], "fixtures")
````
