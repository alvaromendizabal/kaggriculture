## 2 · Bounded execution

The screen checks every one of 527 known control-path observations. If actions are all identical, an explicit full-path identity proof avoids another simulation. Otherwise, only one responsive continuation is run. Each stage has a 120-second worker cap and ten-second heartbeats. The notebook wrapper also terminates the supervisor on interruption or an emergency deadline; the supervisor cleans up its child.

```python
def run_stage(stage):
    if stage not in ('screen', 'pilot'):
        raise ValueError('Unregistered stage')
    console = BASE / 'outputs' / (stage + '_notebook_console.txt')
    console.parent.mkdir(exist_ok=True)
    process = subprocess.Popen([sys.executable, str(BASE / 'run_factorial.py'), stage],
        cwd=BASE, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
        text=True, bufsize=1, start_new_session=True)
    expired = threading.Event()
    def emergency_stop():
        expired.set()
        if process.poll() is None:
            os.killpg(process.pid, signal.SIGTERM)
    timer = threading.Timer(132, emergency_stop)
    timer.start()
    try:
        with console.open('a') as handle:
            for line in process.stdout:
                print(line, end='', flush=True)
                handle.write(line); handle.flush()
        result = process.wait(timeout=4)
        if expired.is_set() or result:
            raise RuntimeError('Stage stopped. Save the notebook and run the bundle command; do not retry unchanged.')
    except BaseException:
        if process.poll() is None:
            os.killpg(process.pid, signal.SIGTERM)
            try:
                process.wait(timeout=4)
            except subprocess.TimeoutExpired:
                os.killpg(process.pid, signal.SIGKILL)
                process.wait()
        raise
    finally:
        timer.cancel()
```
