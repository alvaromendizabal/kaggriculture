Full research implementation is public in this repository. This entry-point notebook presents supplied aggregate evidence; it does not execute agents, rerun screens, or claim a leaderboard result. Negative findings and evidence boundaries are part of the research story.

```python
from pathlib import Path
import json
from datetime import datetime, timezone
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default = 'plotly_mimetype+notebook'
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'research/PUBLICATION_MANIFEST.json').is_file())
DATA = ROOT / 'docs/research/data'
summary = pd.read_csv(DATA / 'screen_summary.csv')
groups = pd.read_csv(DATA / 'group_summary.csv')
latency = pd.read_csv(DATA / 'callback_summary.csv')
hours = pd.read_csv(DATA / 'daypart_summary.csv')
display(summary)
figures = []
```

## Decision activation
A changed action establishes that a representation reaches the decision rule. It does not establish a better outcome.

```python
fig = px.bar(summary, x='family', y='changed_actions', title='Decision changes in reviewed development screens', labels={'changed_actions':'Changed observations','family':'Feature family'})
fig.show(); figures.append(fig)
```

```python
groups['fraction'] = groups.changed_actions / groups.observations
fig = px.bar(groups, x='development_group', y='fraction', color='family', pattern_shape='seat', barmode='group', title='Activation by recorded development group and seat', labels={'fraction':'Changed-observation fraction'})
fig.show(); figures.append(fig)
```

```python
hours['fraction'] = hours.changed_actions / hours.observations
fig = px.bar(hours, x='daypart', y='fraction', color='family', barmode='group', title='Decision activation across the daily cycle', labels={'fraction':'Changed-observation fraction'})
fig.show(); figures.append(fig)
```

## Execution evidence
Timing and correctness checks describe the screened execution. Neither is proof of generalization or competitive superiority.

```python
fig = px.bar(latency, x='statistic', y='milliseconds', color='family', barmode='group', title='Complete callback latency in archived screens', labels={'milliseconds':'Milliseconds'})
fig.show(); figures.append(fig)
```

```python
fig = px.bar(summary, x='family', y='screen_seconds', title='Measured screen duration', labels={'screen_seconds':'Seconds','family':'Feature family'})
fig.show(); figures.append(fig)
```

```python
fig = px.bar(summary, x='family', y='reported_tests_passed', title='Tests reported as passed per family (shared tests overlap)', labels={'reported_tests_passed':'Passing tests in each family command','family':'Feature family'})
fig.show(); figures.append(fig)
```

## Interpretation and next decision
Both screens completed, but zero endpoint comparisons were completed in the source evidence. The next step is a matched game-outcome test, followed by broader independent validation. No official Kaggle rating is claimed.

```python
OUT = ROOT / 'docs/research'
OUT.mkdir(parents=True, exist_ok=True)
html = '<!doctype html><html><head><meta charset="utf-8"><title>Kaggriculture research evidence</title></head><body><main style="max-width:1150px;margin:40px auto;font-family:system-ui"><h1>Kaggriculture — feature research evidence</h1><p>Alvaro Mendizabal · Archived development screening, not leaderboard results. Implementation and tests are public in the repository.</p>'
for i, fig in enumerate(figures):
    html += fig.to_html(full_html=False, include_plotlyjs=True if i == 0 else False)
html += '</main></body></html>'
(OUT / 'index.html').write_text(html, encoding='utf-8')
(OUT / 'overview_execution.json').write_text(json.dumps({'utc': datetime.now(timezone.utc).isoformat(), 'figures': len(figures), 'scope': 'reporting on supplied aggregate evidence', 'new_games': 0}, indent=2))
print('PUBLIC_OVERVIEW_RENDERED', len(figures), 'figures; save and reopen this notebook.')
```
