```python
from pathlib import Path
import datetime as dt
import json
import sys
import time
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import display, Markdown

BASE = Path.cwd()
if not (BASE / "verified_evidence.json").exists():
    candidate = Path.home() / "kaggriculture_manual_resume"
    if not (candidate / "verified_evidence.json").exists():
        raise RuntimeError("Open this notebook from the extracted recovery-package directory")
    BASE = candidate
E = json.loads((BASE / "verified_evidence.json").read_text())
READINESS = BASE / "state/readiness.json"
MODE = "WORKSPACE-VERIFIED" if READINESS.exists() else "REFERENCE-ONLY"
STARTED = dt.datetime.now(dt.timezone.utc).isoformat()
if MODE == "WORKSPACE-VERIFIED":
    sys.path.insert(0, str(BASE))
    import resume
    runtime = json.loads((BASE / "state/runtime.json").read_text())
    assert Path(sys.executable).absolute() == Path(runtime["executable"]).absolute(), \
        "Wrong kernel: choose Kaggriculture Manual (verified source)"
    REPO = Path(runtime["repo"])
    exec(resume.runtime_probe(REPO), {})
    resume.DEADLINE = time.monotonic() + 60
    resume.audit()
    readiness = json.loads(READINESS.read_text())
    B = json.loads((REPO / "reports/bottleneck_research.json").read_text())
    W = json.loads((REPO / "reports/workforce_scope_audit.json").read_text())
else:
    readiness = None
    B, W = E["bottlenecks"], E["workforce"]

OUT = BASE / "outputs"
OUT.mkdir(exist_ok=True)
pio.renderers.default = "plotly_mimetype"
figures = []
display(Markdown(f"## {MODE} MODE\n\n**Started UTC:** {STARTED}  \
**New games:** 0 · **Model fits:** 0 · **Official score improvement measured:** No"))
```

## 3 · Feature coverage is not feature utility

The staffing and later bottleneck audits reused **161 saved observations**. Variation
shows that a descriptor was activated in that slice—not that it improves decisions.
Constants can indicate insufficient state coverage. These family counts are not a
claim about the unique union of every project feature.

```python
fig = go.Figure()
fig.add_bar(name="Varying in saved slice", x=["Staffing", "Bottleneck"],
            y=[S["varying_features"], B["varying_columns"]])
fig.add_bar(name="Constant in saved slice", x=["Staffing", "Bottleneck"],
            y=[S["constant_features"], B["constant_columns"]])
fig.update_layout(title="Candidate coverage—not selected predictive features",
                  barmode="stack", yaxis_title="Candidate descriptors", height=430)
fig.show(); figures.append(("feature_coverage", fig))
display(pd.DataFrame([{"Saved observations": B["observations"],
                       "Base-assignment parity states": B["base_assignment_parity_states"],
                       "High-Spearman pairs": B["high_spearman_pairs"],
                       "New games in that audit": B["new_games"],
                       "Official metric effect measured": B["official_metric_effect_measured"]}]))
```

## 5 · Do not confuse a fast component with a passing full callback

These are the published timings of the saved-observation bottleneck extractor,
**including its route menu**, not a complete deployed agent callback. The original
failed callback must not be replaced by one of these smaller measurements.

```python
latency = B["latency_including_menu_ms"]
fig = go.Figure(go.Bar(x=["Median", "95th percentile", "Maximum"],
                       y=[latency["median"], latency["p95"], latency["max"]],
                       text=[f"{latency[k]:.2f} ms" for k in ["median", "p95", "max"]],
                       textposition="outside"))
fig.update_layout(title="Saved-state feature + menu latency (not full-callback acceptance)",
                  yaxis_title="Milliseconds", height=430)
fig.show(); figures.append(("component_latency", fig))
```
