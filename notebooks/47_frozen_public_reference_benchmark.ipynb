# Kaggriculture — frozen public-reference benchmark
## Research question
Do either of the two exact captured public controllers outperform our deployed controller in **closed-loop games on the same pinned official engine**? This is a reference comparison, not a new original strategy, reproduced leaderboard rating, or submission-ready conclusion.

**Why this step:** crop-program substitution improved our own cash but raised opponent cash more. The newly captured Farm write-up reports its own failed crop overlays and identifies limited worker capacity. We therefore measure the complete references before extracting mechanisms or constructing another overlay.

### Provenance and missing capabilities
The three programs are the deployed sale-timing export, Thomas Tschinkel's Farm export, and the newly captured current Tetsutani timing export. The Farm export matches its publisher's hash after reproducing the `%%writefile` final-newline convention; the new timing export is recovered from its hash-verified embedded archive. Original code and notices remain unchanged. The original notebooks are never executed.

The Farm author's 96% benchmark and historical rating are **author-reported, not our results**. Its reported top-team losses and failed overlays are relevant counterevidence. The source snapshots do not include a complete upstream plan optimizer. Optional `V92_SELL_LIB` is not available and is not injected; its associated optional branch is inactive.

### Evaluation design
Four smoke games precede a 96-game round robin: three agent pairs, 16 unused fixed seeds, two seats. Every game starts new agent interpreters and advances the official game state with both agents' actual actions. Observations contain only the actor's private state plus public fields. Seeds remain evaluator-only.

The unit of uncertainty is the **seed**, not the seat. Within each pair, both seats are averaged before a 10,000-replicate seed bootstrap. For each of two reference-versus-incumbent comparisons, nomination requires score ≥ 0.70, each seat ≥ 0.625, positive cash-margin bootstrap lower bound, and a one-sided exact seed sign-test p ≤ 0.025. Tied or losing clusters are not counted as positive sign-test successes. Nomination means another untouched confirmation is justified; it never triggers a submission. The reference-versus-reference match is exploratory.

### Important comparability limits
Sixteen prespecified seeds are a small engineering holdout, not a probability sample of the ladder. A common initial seed does not guarantee identical future shop draws when strategies alter RNG consumption. Actual shop sequences are retained; they are not manually forced to match. Terminal cash differences are secondary to win/tie outcomes. No historical author result is silently mixed into this evaluation.

## Closed-loop win–tie outcomes

```python
fig=go.Figure()
for c in COMPARISONS:
    lo,hi=c['win_tie_95_seed_bootstrap']
    fig.add_trace(go.Bar(x=[c['agent']+' vs '+c['opponent']],y=[c['score']],name=c['agent']+' vs '+c['opponent'],error_y=dict(type='data',symmetric=False,array=[max(0,hi-c['score'])],arrayminus=[max(0,c['score']-lo)])))
if not COMPARISONS: pending(fig)
fig.add_hline(y=.5,line_dash='dot')
fig.update_layout(title='Paired reference comparisons — 95% whole-seed bootstrap intervals',yaxis_title='Win–tie score (0 to 1)',xaxis_title='First agent versus second agent',showlegend=False)
fig.update_yaxes(range=[0,1])
show(fig)
```
