## Data, legal features, and target
The frozen study selects up to 64 episodes per retained day. September 16–18 supply training, September 19 validation, and September 20 the untouched model-selection test. Duplicate episodes are assigned to their earliest source day before sampling. Both seats and all sampled choices stay with their episode. This is episode-disjoint chronological evaluation, not a claim of code-lineage-disjoint generalization.

Features use only current public state, own private state, and our proposed orders. The teacher may use the opponent’s recorded simultaneous actions and private state only to reconstruct labels. It first reproduces both actual next cash balances, then swaps one adjacent pair of distinct cash-product SELL orders in a sale-only queue. It cannot change our quantities, staffing, production, purchases, or routes.

Up to 32 eligible decisions per seat are sampled without using outcomes. The sum of their one-turn gains is **not** a full-season return, and the hindsight best swap is only a local action-space upper bound.

## Checkpoint-preserving recovery
The preceding run stopped before fitting after 319 episode receipts were written. The frozen data selection and thresholds are unchanged. Successful episode bodies can be adopted only after source-contract and checksum validation. A training-only action-opportunity check now runs before further reconstruction or fitting. If it is too weak, the direction ends without training.

A remaining 38-coin discrepancy is not ignored, tolerated, or used to tune a model. For a viable training opportunity, a causal reconstruction of private inventory insertion order must reproduce every cash and private-quantity transition in the offending episode. Only a demonstrated order-loss mechanism can enable resumption. An unresolved discrepancy stops training and preserves a structured, order-preserving diagnostic capsule. This mechanism was tested on controlled fixtures; the failed public episode is not described as repaired before its own audit passes.

## Learning and selection
Two fixed multioutput forests predict own-cash and relative-cash changes. They use 96 trees, maximum depth 8, and minimum leaf size 20. Training weights balance episode contributions. Only validation can choose the model; thresholds are fixed prospectively. Test labels are evaluated after the selected model/export hashes are locked. The policy changes an action only when predicted own gain is at least 1 coin and relative gain at least 5 coins.

The training-only hindsight opportunity gate must exceed 25 sampled coins per episode-seat before either fit. A validation/test pass requires positive episode-bootstrap lower bounds, at least 20 episodes and 10 changed episodes, at least 5 sampled relative coins per episode-seat, and at most 25% harmful changed decisions. These gates justify a fresh-game test, not deployment. Forest export parity is checked before any match.

## Limitations and reproducibility
This is a restricted independent market-order component added to an attributed existing controller. It does not reproduce the full upstream multi-day labor optimizer. Policy states induced by repeated interventions can differ from logged states; rival behavior may respond. Neither a one-step gain nor a local win is an official rating. The newest full daily payloads and every retained historical episode are not claimed as training inputs.

The study preserves existing raw files and prior checkpoints. New action-target caches are bound to source hashes, selected episode IDs and implementation identity. Read-only connected inspection is separate from the user-run fitting, matches and S3 artifact writes. External notebook sources retain contributor notices. The source policy and model export remain unchanged between gates.
