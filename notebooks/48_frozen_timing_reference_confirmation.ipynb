# Kaggriculture — frozen timing-reference confirmation
## From exploratory nomination to a controlled release decision
**Objective:** determine whether the exact captured Tetsutani timing controller deserves a single attributed reference submission, without modifying its code or optimizing on confirmation outcomes.

The completed nomination benchmark measured 29 wins and 3 losses against the deployed controller, and 29 wins and 3 losses against the Farm reference. It used 16 seed clusters and cannot substitute for this confirmation. Its 100 games are preserved and excluded from every confirmation estimate below.

### What is and is not original work
The controller is an exact public export, including the original authors' code and Apache notices. Our contribution in this milestone is the source-bound comparative evaluation, uncertainty analysis, isolated execution, reproducible integration and release checks. It is **not** independent authorship of the full controller or recreation of the upstream labor-plan optimizer. The captured version combines a changed productive opening and late seed controls, but this whole-system comparison does not identify their individual causal contributions.

### Locked prospective design
1. **Direct stage:** 32 previously unused seeds, both seats: 64 games against the incumbent. Require win–tie score ≥ 0.75, each seat ≥ 0.625, one-sided seed-level exact binomial p ≤ 0.01, and a positive 95% whole-seed bootstrap lower bound on cash advantage.
2. **Fixed field:** a different set of 32 seeds, three attributed fixed opponents, both seats, both controllers: 384 games. Require nonnegative observed mean win–tie change overall, per opponent and per seat. The whole-seed bootstrap lower bound must be ≥ −0.05. Own cash must not decline on average, relative cash must be nonnegative per opponent and seat, and its overall bootstrap lower bound must be positive.
3. **Export gate:** four games on another seed compare the reference controller with the exact re-extracted submission file. Their actions and physical results must match. These are correctness evidence, not extra performance trials.

The field rule is prospective **non-inferiority within a fixed panel**, with a five-percentage-point tolerance at the interval's lower bound. It is not a claim of universal field superiority. No positive early stopping, threshold changes, rerolled seeds or post-result candidate changes are allowed. Direct and field seed sets are disjoint.

### Important validity limits
Prespecified deterministic seed panels are engineering holdouts, not probability samples of the live ladder. Seats and opponent contexts sharing a seed are clustered together. Exact-test and bootstrap interpretations assume adequate exchangeability of those clusters. A common initial seed need not yield identical town-shop histories after policies alter RNG consumption; actual histories are retained, not forced to match. Winning magnitude is a secondary diagnostic, not a substitute for wins/ties. Neither a local pass nor a submission receipt establishes a new official rating.
