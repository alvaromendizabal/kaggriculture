# 05 · Adaptive synthesis

This executed review notebook presents the bounded development evidence for the authored adaptive overlay. The public teacher routes are external baselines; canonical validation and holdout seeds remain untouched.

```python
import json
from pathlib import Path

import pandas as pd
import plotly.express as px
from IPython.display import display

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
REPORTS = ROOT / "reports"

stage_a = pd.read_csv(REPORTS / "adaptive_stage_a_summary.csv")
stage_b = pd.read_csv(REPORTS / "adaptive_stage_b_by_opponent.csv")
receipt = json.loads((REPORTS / "adaptive_synthesis_receipt.json").read_text())

assert receipt["decision"] == "ADAPTIVE_SYNTHESIS_V1_PROMOTED"
assert receipt["stage_b_wins"] == 24
assert receipt["validation_seeds_used"] is False
assert receipt["holdout_seeds_used"] is False

display(stage_a)
display(stage_b)
print("Stage B mean margin:", round(receipt["stage_b_mean_margin"], 1))
```

```python
fig = px.bar(
    stage_a.sort_values("mean_margin"),
    x="mean_margin",
    y="adaptive_mode",
    orientation="h",
    text="wins",
    title="Stage A · Sparse adaptive overlay ablation",
    labels={"mean_margin": "Mean bank margin", "adaptive_mode": "Variant"},
)
fig.update_layout(width=960, height=430, margin=dict(l=110, r=40, t=80, b=70))
fig.add_vline(x=0, line_dash="dash")
fig.show()
```

```python
fig = px.bar(
    stage_b.sort_values("mean_margin"),
    x="mean_margin",
    y="opponent",
    orientation="h",
    text="wins",
    title="Stage B · Promoted overlay versus external teacher league",
    labels={"mean_margin": "Mean bank margin", "opponent": "External teacher"},
)
fig.update_layout(width=1020, height=470, margin=dict(l=170, r=40, t=80, b=70))
fig.add_vline(x=0, line_dash="dash")
fig.show()
```

`clone_terminal` passed the bounded development gate with 24/24 Stage B wins. This result justifies packaging-equivalence testing and a single controlled temperature submission; it does **not** establish a live leaderboard rating or final-tournament result.
