```python
import json
from pathlib import Path
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.templates.default = "none"
root = Path.cwd()
if root.name == "notebooks": root = root.parent
report = json.loads((root / "reports/recent_meta_frontier.json").read_text())
experiments = pd.DataFrame(report["experiments"])
print({"official_engine_games": report["aggregate_work_since_last_publication"]["accepted_official_engine_games"], "model_fits": report["aggregate_work_since_last_publication"]["model_fits"], "experiments": len(experiments)})
```

```python
rows=[]
for item in report["experiments"]:
    if item.get("screen_score") is not None:
        rows.append((item["name"],"screen",item["screen_score"],item.get("screen_control_score")))
    if item.get("confirmation_score") is not None:
        rows.append((item["name"],"confirmation",item["confirmation_score"],item.get("confirmation_control_score")))
comparison=pd.DataFrame(rows,columns=["experiment","stage","candidate","control"])
fig=go.Figure()
for stage in comparison.stage.unique():
    f=comparison[comparison.stage==stage]
    fig.add_bar(name=f"candidate · {stage}",x=f.experiment,y=f.candidate)
    fig.add_bar(name=f"control · {stage}",x=f.experiment,y=f.control)
fig.update_layout(title="Playable experiments: candidate vs registered control",yaxis_title="family-weighted win/tie score",yaxis_tickformat=".0%",barmode="group",height=480,width=960)
fig.show()
```

```python
f=report["data_and_forecasting"]
frame=pd.DataFrame([
    ("final-date test",f["cash_forecast_test"]["mae"],f["cash_forecast_test"]["persistence_mae"]),
    ("active-ladder transfer",f["ladder_transfer"]["mae"],f["ladder_transfer"]["persistence_mae"]),
],columns=["evaluation","forecast","baseline"])
fig=go.Figure()
fig.add_bar(name="frozen forecast",x=frame.evaluation,y=frame.forecast)
fig.add_bar(name="persistence baseline",x=frame.evaluation,y=frame.baseline)
fig.update_layout(title="Auxiliary forecast transferred to later ladder observations",yaxis_title="mean absolute error · lower is better",barmode="group",height=420,width=820)
fig.show()
print("Active scoring pair:", [(x["submission_ref"],x["rating"]) for x in report["active_scoring_pair"]])
```

## Research conclusion

- **Forecasting transferred, but prediction quality was not treated as game-performance evidence.** The frozen public-state model reduced cash-change MAE by about 26% on both its final-date test and later active-ladder observations.
- **Narrow playable overlays did not generalize strongly enough.** Learned supply execution weakened on independent confirmation, whole-order queue planning regressed, and contextual production-program selection produced only a small confirmation gain.
- **The strongest recent fixed-panel result came from an attributed external whole-controller reference.** It reached 94.44% family-weighted confirmation score versus 85.42% for labor, but the paired seed-level uncertainty interval still crossed zero; it therefore was not promoted under the registered gate.

The next registered step is one storage-safe fresh-seed replication of that exact unchanged reference. The previous rejected confirmation is not pooled into the new verdict. A passing replication permits one gated terminal submission; a failure closes the reference without automatic sample expansion.
