## Mechanism and boundaries

The component starts at decision step 696. It searches eight candidate crew/route combinations: eight through eleven hired hands and two deterministic task orderings. A deadline-constrained insertion heuristic assigns required harvests, immediate water bonuses and return-to-shed actions. A width-four per-product beam search coordinates observed-price sales with predicted own deliveries and shed capacity.

The forecast assumes **zero future rival transactions**, then re-plans sales using each newly observed market. This makes its modeled cash value a planning heuristic, **not a promised outcome or an unbiased causal estimate**. It reads only current legal observations and own history. The existing reference is used when a canonical feasible final-day plan is unavailable. This pilot tests the combined intervention; it does not identify separate causal contributions of staffing, routing and sales.
