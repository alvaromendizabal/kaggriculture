This notebook summarizes the latest Kaggriculture milestone from the canonical AWS/SageMaker workflow. It intentionally separates **local official-engine evaluation** from **live Kaggle rating**.

The key result is not a leaderboard win. A narrowly targeted labor-funding intervention passed two substantial local evaluations, then collapsed on the live ladder. That divergence is now treated as evidence that the local opponent league is not representative enough for promotion decisions.

**Public boundary:** raw episode archives, private caches, submission tarballs, credentials, and third-party controller source are intentionally excluded.

```python
import pandas as pd
import plotly.graph_objects as go

local = pd.DataFrame(
    [
        {
            "evaluation": "Independent confirmation",
            "policy": "Labor treasury",
            "score": 0.86875,
            "games": 480,
        },
        {
            "evaluation": "Independent confirmation",
            "policy": "Race-up control",
            "score": 0.7770833333,
            "games": 480,
        },
        {
            "evaluation": "Independent confirmation",
            "policy": "Parent control",
            "score": 0.7416666667,
            "games": 480,
        },
        {
            "evaluation": "Champion comparison",
            "policy": "Labor treasury",
            "score": 0.85,
            "games": 240,
        },
        {
            "evaluation": "Champion comparison",
            "policy": "Race-up control",
            "score": 0.75,
            "games": 240,
        },
        {
            "evaluation": "Champion comparison",
            "policy": "Timing reference",
            "score": 0.5083333333,
            "games": 240,
        },
    ]
)
live = pd.DataFrame(
    [
        {"submission": "Timing reference", "rating": 2048.7, "ref": "56471163"},
        {"submission": "Original sale timing", "rating": 1940.7, "ref": "56441483"},
        {"submission": "Race-up", "rating": 1851.5, "ref": "56511669"},
        {"submission": "Labor treasury", "rating": 600.0, "ref": "56598411"},
    ]
)

print("Local evaluation rows:", len(local))
display(local)
```

```python
fig = go.Figure()
for evaluation in local["evaluation"].unique():
    frame = local[local["evaluation"] == evaluation]
    fig.add_bar(
        name=evaluation,
        x=frame["policy"],
        y=frame["score"],
        text=[f"{v:.1%}" for v in frame["score"]],
        textposition="outside",
    )
fig.update_layout(
    title="Local official-engine win/tie score",
    yaxis_title="Win/tie score",
    yaxis_tickformat=".0%",
    barmode="group",
    height=480,
    width=900,
    legend_title="Evaluation",
)
fig.show()
```

```python
fig = go.Figure(
    go.Bar(
        x=live["submission"],
        y=live["rating"],
        text=[f"{v:.1f}" for v in live["rating"]],
        textposition="outside",
    )
)
fig.update_layout(
    title="Returned completed Kaggle ratings",
    yaxis_title="Kaggle rating",
    height=460,
    width=900,
)
fig.show()
```

## What the local experiments established

The labor-treasury policy was intentionally sparse: it changed the existing race-up controller only when next-morning labor funding was at risk. In independent confirmation it reached **86.875%** win/tie score versus **77.708%** for race-up and **74.167%** for the parent. A later 720-game champion comparison again favored labor (**85.0%**) over race-up (**75.0%**) and the timing reference (**50.83%**).

The champion comparison used **12 fresh seed clusters × 10 opponents × both seats**, for 240 games per policy. Against race-up, the observed gain was **+10 percentage points**, driven by a complete reversal against Shop Router V7 while other matched opponent results were unchanged. The retrospective best-of-fixed-policies upper bound was only **+2.08 percentage points** over the labor policy, so a simple policy-selector ensemble had little additional local headroom.

## What live feedback disproved

The submitted labor archive, reference **56598411**, subsequently completed with a returned rating of **600.0**. The strongest completed submission in the same authenticated history remained the frozen timing reference at **2048.7**.

The candidate archive passed local source/loader checks before submission, but the installed CLI did not expose a verified submission-download interface, so server-side archive-byte identity could not be independently re-downloaded and hashed after upload.

**Conclusion:** the local league was useful for controlled mechanism testing but was not sufficiently representative to predict live ladder performance. The live result invalidates using this local score alone as a promotion instrument.

## Next research priority

The next ceiling-escape step is **validation redesign before further policy tuning**:

1. refresh only new public/top-agent episode evidence;
2. build a current-meta opponent panel with stronger diversity and held-out families;
3. measure which local metrics actually correlate with live outcomes;
4. add promotion gates based on recent-opponent robustness rather than aggregate static win rate alone;
5. only then resume structural policy search (on-policy strategic control / league training) against the improved evaluator.

This keeps the project focused on the real failure mode exposed by the live submission rather than optimizing harder against a misleading local benchmark.
